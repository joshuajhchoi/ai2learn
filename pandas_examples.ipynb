## 2. 판다스 자료구조 이해

### 2.1 Series

```python
# Series 생성
# 리스트로부터 Series 생성
s1 = pd.Series([1, 2, 3, 4, 5])
print("리스트로부터 Series 생성:")
print(s1)

# 인덱스 지정하여 Series 생성
s2 = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
print("\n인덱스 지정하여 Series 생성:")
print(s2)

# 딕셔너리로부터 Series 생성
data_dict = {'a': 10, 'b': 20, 'c': 30, 'd': 40, 'e': 50}
s3 = pd.Series(data_dict)
print("\n딕셔너리로부터 Series 생성:")
print(s3)
```

```python
# Series 속성 및 메서드
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print("Series:")
print(s)

# 값 배열
print("\n값 배열 (values):", s.values)

# 인덱스
print("인덱스 (index):", s.index)

# 데이터 타입
print("데이터 타입 (dtype):", s.dtype)

# 크기
print("크기 (size):", s.size)

# 기본 통계
print("\n기본 통계 (describe):")
print(s.describe())
```

```python
# Series 인덱싱 및 슬라이싱
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print("Series:")
print(s)

# 위치 기반 인덱싱
print("\n첫 번째 요소:", s[0])
print("마지막 요소:", s[4])

# 라벨 기반 인덱싱
print("\n라벨 'a'의 값:", s['a'])
print("라벨 'e'의 값:", s['e'])

# 여러 라벨 선택
print("\n여러 라벨 선택:")
print(s[['a', 'c', 'e']])

# 슬라이싱 (위치 기반)
print("\n위치 기반 슬라이싱 (처음 3개):")
print(s[:3])

# 슬라이싱 (라벨 기반)
print("\n라벨 기반 슬라이싱 ('b'부터 'd'까지):")
print(s['b':'d'])
```

```python
# Series 연산
s1 = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
s2 = pd.Series([5, 10, 15, 20, 25], index=['a', 'b', 'c', 'd', 'e'])
print("Series 1:")
print(s1)
print("\nSeries 2:")
print(s2)

# 덧셈
print("\n덧셈 (s1 + s2):")
print(s1 + s2)

# 뺄셈
print("\n뺄셈 (s1 - s2):")
print(s1 - s2)

# 곱셈
print("\n곱셈 (s1 * s2):")
print(s1 * s2)

# 나눗셈
print("\n나눗셈 (s1 / s2):")
print(s1 / s2)

# 스칼라 연산
print("\ns1 * 2:")
print(s1 * 2)
```

```python
# 인덱스가 다른 Series 연산
s1 = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
s2 = pd.Series([5, 10, 15, 20, 25], index=['a', 'c', 'e', 'g', 'i'])
print("Series 1:")
print(s1)
print("\nSeries 2:")
print(s2)

# 덧셈 (공통 인덱스만 연산, 나머지는 NaN)
print("\n덧셈 (s1 + s2):")
print(s1 + s2)

# NaN 처리
print("\n덧셈 후 NaN을 0으로 채우기:")
print((s1 + s2).fillna(0))
```

### 2.2 DataFrame

```python
# DataFrame 생성
# 딕셔너리로부터 DataFrame 생성
data = {
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '나이': [25, 28, 22, 30, 27],
    '성별': ['남', '여', '남', '여', '남'],
    '직업': ['학생', '회사원', '학생', '연구원', '회사원']
}

df1 = pd.DataFrame(data)
print("딕셔너리로부터 DataFrame 생성:")
print(df1)

# 리스트의 리스트로부터 DataFrame 생성
data_list = [
    ['김철수', 25, '남', '학생'],
    ['이영희', 28, '여', '회사원'],
    ['박민수', 22, '남', '학생'],
    ['정지은', 30, '여', '연구원'],
    ['최동욱', 27, '남', '회사원']
]
columns = ['이름', '나이', '성별', '직업']

df2 = pd.DataFrame(data_list, columns=columns)
print("\n리스트의 리스트로부터 DataFrame 생성:")
print(df2)

# Series 딕셔너리로부터 DataFrame 생성
s1 = pd.Series(['김철수', '이영희', '박민수', '정지은', '최동욱'])
s2 = pd.Series([25, 28, 22, 30, 27])
s3 = pd.Series(['남', '여', '남', '여', '남'])
s4 = pd.Series(['학생', '회사원', '학생', '연구원', '회사원'])

df3 = pd.DataFrame({'이름': s1, '나이': s2, '성별': s3, '직업': s4})
print("\nSeries 딕셔너리로부터 DataFrame 생성:")
print(df3)
```

```python
# DataFrame 속성 및 메서드
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '나이': [25, 28, 22, 30, 27],
    '성별': ['남', '여', '남', '여', '남'],
    '직업': ['학생', '회사원', '학생', '연구원', '회사원']
})

print("DataFrame:")
print(df)

# 기본 정보
print("\n기본 정보 (info):")
df.info()

# 컬럼명
print("\n컬럼명 (columns):", df.columns)

# 인덱스
print("인덱스 (index):", df.index)

# 데이터 타입
print("\n데이터 타입 (dtypes):")
print(df.dtypes)

# 크기
print("\n크기 (shape):", df.shape)

# 차원
print("차원 (ndim):", df.ndim)

# 요소 개수
print("요소 개수 (size):", df.size)

# 기본 통계
print("\n기본 통계 (describe):")
print(df.describe())

# 처음 n행
print("\n처음 2행 (head):")
print(df.head(2))

# 마지막 n행
print("\n마지막 2행 (tail):")
print(df.tail(2))
```

## 4. 데이터 조회 및 선택

```python
# 샘플 데이터 생성
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '나이': [25, 28, 22, 30, 27],
    '성별': ['남', '여', '남', '여', '남'],
    '직업': ['학생', '회사원', '학생', '연구원', '회사원'],
    '점수': [85, 92, 78, 96, 88]
})

print("DataFrame:")
print(df)
```

```python
# 열 선택
# 단일 열 선택 (Series 반환)
print("'이름' 열 선택 (대괄호):")
print(df['이름'])

print("\n'이름' 열 선택 (점 표기법):")
print(df.이름)

# 여러 열 선택 (DataFrame 반환)
print("\n'이름'과 '나이' 열 선택:")
print(df[['이름', '나이']])
```

```python
# 행 선택
# 위치 기반 행 선택 (슬라이싱)
print("처음 3행 선택:")
print(df[:3])

# iloc: 위치 기반 인덱싱
print("\n첫 번째 행 선택 (iloc):")
print(df.iloc[0])

print("\n2~4번째 행 선택 (iloc):")
print(df.iloc[1:4])

# loc: 라벨 기반 인덱싱
print("\n인덱스 2인 행 선택 (loc):")
print(df.loc[2])
```

```python
# 행과 열 동시 선택
# iloc: 위치 기반 인덱싱
print("첫 번째 행, 두 번째 열 선택 (iloc):")
print(df.iloc[0, 1])

print("\n처음 2행, 처음 3열 선택 (iloc):")
print(df.iloc[:2, :3])

# loc: 라벨 기반 인덱싱
print("\n인덱스 1인 행, '이름'과 '나이' 열 선택 (loc):")
print(df.loc[1, ['이름', '나이']])

print("\n인덱스 0~2인 행, '이름'부터 '직업'까지 열 선택 (loc):")
print(df.loc[0:2, '이름':'직업'])
```

```python
# 조건에 따른 필터링
# 단일 조건
print("나이가 25 이상인 행:")
print(df[df['나이'] >= 25])

print("\n성별이 '여'인 행:")
print(df[df['성별'] == '여'])

# 복합 조건
print("\n나이가 25 이상이고 성별이 '남'인 행:")
print(df[(df['나이'] >= 25) & (df['성별'] == '남')])

print("\n직업이 '학생'이거나 점수가 90 이상인 행:")
print(df[(df['직업'] == '학생') | (df['점수'] >= 90)])
```

```python
# query 메서드 사용
print("나이가 25 이상인 행 (query):")
print(df.query('나이 >= 25'))

print("\n직업이 '학생'인 행 (query):")
print(df.query('직업 == "학생"'))

print("\n나이가 25 이상이고 성별이 '남'인 행 (query):")
print(df.query('나이 >= 25 and 성별 == "남"'))
```

```python
# isin 메서드 사용
print("직업이 '학생' 또는 '회사원'인 행:")
print(df[df['직업'].isin(['학생', '회사원'])])

print("\n이름이 '김철수' 또는 '정지은'인 행:")
print(df[df['이름'].isin(['김철수', '정지은'])])
```

## 5. 데이터 전처리 및 가공

```python
# 결측치가 있는 샘플 데이터 생성
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱', '홍길동'],
    '나이': [25, 28, None, 30, 27, None],
    '성별': ['남', '여', '남', None, '남', '남'],
    '직업': ['학생', '회사원', '학생', '연구원', None, '회사원'],
    '점수': [85, 92, 78, None, 88, 76]
})

print("결측치가 있는 DataFrame:")
print(df)
```

```python
# 결측치 확인
print("결측치 확인 (isnull):")
print(df.isnull())

print("\n결측치 개수 (열별):")
print(df.isnull().sum())

print("\n결측치 개수 (행별):")
print(df.isnull().sum(axis=1))

print("\n결측치가 있는 행:")
print(df[df.isnull().any(axis=1)])
```

```python
# 결측치 처리
# 결측치 제거
print("결측치가 있는 행 제거:")
print(df.dropna())

print("\n모든 값이 결측치인 행만 제거:")
print(df.dropna(how='all'))

print("\n특정 열('나이', '점수')에 결측치가 있는 행 제거:")
print(df.dropna(subset=['나이', '점수']))

print("\n결측치가 있는 열 제거:")
print(df.dropna(axis=1))
```

```python
# 결측치 채우기
# 특정 값으로 채우기
print("결측치를 특정 값으로 채우기:")
print(df.fillna({'나이': 0, '성별': '미상', '직업': '무직', '점수': 0}))

# 평균값으로 채우기
print("\n'나이'와 '점수' 열의 결측치를 평균값으로 채우기:")
df_mean_filled = df.copy()
df_mean_filled['나이'] = df_mean_filled['나이'].fillna(df_mean_filled['나이'].mean())
df_mean_filled['점수'] = df_mean_filled['점수'].fillna(df_mean_filled['점수'].mean())
print(df_mean_filled)

# 앞/뒤 값으로 채우기
print("\n앞의 값으로 채우기 (ffill):")
print(df.fillna(method='ffill'))

print("\n뒤의 값으로 채우기 (bfill):")
print(df.fillna(method='bfill'))
```

```python
# 중복 데이터 처리
# 중복 데이터가 있는 샘플 생성
df_dup = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '김철수', '이영희'],
    '나이': [25, 28, 22, 25, 28],
    '성별': ['남', '여', '남', '남', '여']
})

print("중복 데이터가 있는 DataFrame:")
print(df_dup)

# 중복 확인
print("\n중복 행 확인:")
print(df_dup.duplicated())

# 중복 제거
print("\n중복 행 제거:")
print(df_dup.drop_duplicates())

# 특정 열 기준 중복 제거
print("\n'이름' 열 기준 중복 제거:")
print(df_dup.drop_duplicates(subset=['이름']))

# 마지막 값 유지
print("\n중복 시 마지막 값 유지:")
print(df_dup.drop_duplicates(keep='last'))
```

```python
# 데이터 타입 변환
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '나이': ['25', '28', '22', '30', '27'],  # 문자열 타입
    '성별': ['남', '여', '남', '여', '남'],
    '점수': ['85.5', '92.0', '78.5', '96.5', '88.0'],  # 문자열 타입
    '합격여부': ['True', 'True', 'False', 'True', 'True']  # 문자열 타입
})

print("원본 DataFrame:")
print(df)
print("\n데이터 타입:")
print(df.dtypes)

# 데이터 타입 변환
df['나이'] = df['나이'].astype(int)
df['점수'] = df['점수'].astype(float)
df['합격여부'] = df['합격여부'].astype(bool)

print("\n타입 변환 후 DataFrame:")
print(df)
print("\n데이터 타입:")
print(df.dtypes)
```

```python
# 컬럼 조작
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '나이': [25, 28, 22, 30, 27],
    '성별': ['남', '여', '남', '여', '남'],
    '국어': [85, 92, 78, 96, 88],
    '영어': [92, 88, 95, 85, 90],
    '수학': [78, 95, 92, 90, 85]
})

print("원본 DataFrame:")
print(df)

# 컬럼 추가
df['평균'] = (df['국어'] + df['영어'] + df['수학']) / 3
print("\n'평균' 컬럼 추가:")
print(df)

# 컬럼 삭제
df_drop = df.drop(columns=['국어', '영어', '수학'])
print("\n과목 컬럼 삭제:")
print(df_drop)

# 컬럼 이름 변경
df_rename = df.rename(columns={'이름': '성명', '나이': '연령'})
print("\n컬럼 이름 변경:")
print(df_rename)
```

```python
# apply, map, applymap 함수
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '나이': [25, 28, 22, 30, 27],
    '국어': [85, 92, 78, 96, 88],
    '영어': [92, 88, 95, 85, 90],
    '수학': [78, 95, 92, 90, 85]
})

print("원본 DataFrame:")
print(df)

# apply: 행 또는 열에 함수 적용
# 각 행의 국어, 영어, 수학 점수 평균 계산
df['평균'] = df[['국어', '영어', '수학']].apply(np.mean, axis=1)
print("\n각 행의 평균 점수 (apply):")
print(df)

# 각 과목의 평균 점수 계산
subject_means = df[['국어', '영어', '수학']].apply(np.mean, axis=0)
print("\n각 과목의 평균 점수 (apply):")
print(subject_means)

# 사용자 정의 함수 적용
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    else:
        return 'D'

# map: Series의 각 요소에 함수 적용
df['국어등급'] = df['국어'].map(grade)
print("\n국어 등급 추가 (map):")
print(df)

# 딕셔너리를 사용한 매핑
gender_map = {'남': 'Male', '여': 'Female'}
df['성별'] = ['남', '여', '남', '여', '남']
df['Gender'] = df['성별'].map(gender_map)
print("\n성별 영문 변환 (map):")
print(df[['성별', 'Gender']])

# applymap: DataFrame의 모든 요소에 함수 적용
# 숫자 데이터만 있는 DataFrame 생성
df_num = df[['국어', '영어', '수학']]
print("\n숫자 데이터:")
print(df_num)

# 모든 요소를 반올림
df_rounded = df_num.applymap(lambda x: round(x / 10) * 10)
print("\n10단위 반올림 (applymap):")
print(df_rounded)
```

```python
# 그룹별 집계
# 학년별 집계
print("학년별 평균 점수:")
print(df.groupby('학년')[['국어', '영어', '수학']].mean())

# 성별 집계
print("\n성별 평균 점수:")
print(df.groupby('성별')[['국어', '영어', '수학']].mean())

# 여러 열 기준 그룹화
print("\n학년 및 성별 평균 점수:")
print(df.groupby(['학년', '성별'])[['국어', '영어', '수학']].mean())
```

## 8. 데이터 결합 및 병합

```python
# 샘플 데이터 생성
# 학생 정보 DataFrame
students = pd.DataFrame({
    '학번': [1001, 1002, 1003, 1004, 1005],
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '학과': ['컴퓨터공학', '경영학', '컴퓨터공학', '물리학', '경영학']
})

# 성적 정보 DataFrame
scores = pd.DataFrame({
    '학번': [1001, 1002, 1003, 1005, 1006],
    '국어': [85, 92, 78, 88, 95],
    '영어': [92, 88, 95, 90, 85],
    '수학': [78, 95, 92, 85, 90]
})

print("학생 정보:")
print(students)
print("\n성적 정보:")
print(scores)
```

```python
# 데이터 연결 (concatenate)
# 행 방향으로 연결
students1 = students.iloc[:3]
students2 = students.iloc[3:]
print("학생 정보 (앞부분):")
print(students1)
print("\n학생 정보 (뒷부분):")
print(students2)

students_concat = pd.concat([students1, students2])
print("\n행 방향으로 연결:")
print(students_concat)

# 열 방향으로 연결
info1 = students[['학번', '이름']]
info2 = students[['학과']]
print("\n기본 정보:")
print(info1)
print("\n학과 정보:")
print(info2)

info_concat = pd.concat([info1, info2], axis=1)
print("\n열 방향으로 연결:")
print(info_concat)
```

```python
# 데이터 병합 (merge)
# 내부 조인 (inner join)
inner_merge = pd.merge(students, scores, on='학번')
print("내부 조인 (inner join):")
print(inner_merge)

# 외부 조인 (outer join)
outer_merge = pd.merge(students, scores, on='학번', how='outer')
print("\n외부 조인 (outer join):")
print(outer_merge)

# 왼쪽 조인 (left join)
left_merge = pd.merge(students, scores, on='학번', how='left')
print("\n왼쪽 조인 (left join):")
print(left_merge)

# 오른쪽 조인 (right join)
right_merge = pd.merge(students, scores, on='학번', how='right')
print("\n오른쪽 조인 (right join):")
print(right_merge)
```

```python
# 여러 키를 기준으로 병합
# 추가 샘플 데이터 생성
students_courses = pd.DataFrame({
    '학번': [1001, 1001, 1002, 1002, 1003, 1003],
    '과목코드': ['CS101', 'CS102', 'BA101', 'BA102', 'CS101', 'PH101'],
    '학점': ['A', 'B+', 'A+', 'B', 'C+', 'A']
})

courses = pd.DataFrame({
    '과목코드': ['CS101', 'CS102', 'BA101', 'BA102', 'PH101'],
    '과목명': ['프로그래밍기초', '자료구조', '경영학원론', '회계원리', '일반물리학'],
    '학점수': [3, 3, 2, 2, 3]
})

print("학생 수강 정보:")
print(students_courses)
print("\n과목 정보:")
print(courses)

# 학번과 과목코드를 기준으로 학생 정보와 수강 정보 병합
result1 = pd.merge(students, students_courses, on='학번')
print("\n학생 정보와 수강 정보 병합:")
print(result1)

# 과목코드를 기준으로 수강 정보와 과목 정보 병합
result2 = pd.merge(result1, courses, on='과목코드')
print("\n최종 병합 결과:")
print(result2)
```

```python
# 조인 (join)
# 인덱스를 기준으로 조인
students_indexed = students.set_index('학번')
scores_indexed = scores.set_index('학번')
print("인덱스가 학번인 학생 정보:")
print(students_indexed)
print("\n인덱스가 학번인 성적 정보:")
print(scores_indexed)

# join 메서드 사용
joined = students_indexed.join(scores_indexed)
print("\n인덱스 기준 조인 (기본: left join):")
print(joined)

# 다른 조인 방식 지정
joined_inner = students_indexed.join(scores_indexed, how='inner')
print("\n인덱스 기준 내부 조인:")
print(joined_inner)
```

## 10. 시계열 및 텍스트 데이터 처리

```python
# 시계열 데이터 생성
date_range = pd.date_range(start='2023-01-01', end='2023-12-31', freq='M')
print("월별 날짜 범위:")
print(date_range)

# 시계열 데이터프레임 생성
np.random.seed(42)
ts_data = pd.DataFrame({
    '날짜': date_range,
    '매출': np.random.randint(1000, 5000, size=len(date_range)),
    '비용': np.random.randint(800, 3000, size=len(date_range)),
    '고객수': np.random.randint(100, 500, size=len(date_range))
})

# 날짜를 인덱스로 설정
ts_data.set_index('날짜', inplace=True)
print("\n시계열 데이터:")
print(ts_data)
```

```python
# 시계열 데이터 인덱싱 및 슬라이싱
# 특정 날짜 선택
print("2023년 3월 데이터:")
print(ts_data.loc['2023-03-31'])

# 날짜 범위 선택
print("\n2023년 상반기 데이터:")
print(ts_data.loc['2023-01-31':'2023-06-30'])

# 날짜 속성 추출
ts_data['연도'] = ts_data.index.year
ts_data['월'] = ts_data.index.month
ts_data['분기'] = ts_data.index.quarter
print("\n날짜 속성 추가:")
print(ts_data)
```

```python
# 시계열 데이터 리샘플링
# 분기별 리샘플링
quarterly = ts_data[['매출', '비용', '고객수']].resample('Q').mean()
print("분기별 평균:")
print(quarterly)

# 반기별 리샘플링
half_yearly = ts_data[['매출', '비용', '고객수']].resample('6M').sum()
print("\n반기별 합계:")
print(half_yearly)
```

```python
# 시계열 데이터 시각화
plt.figure(figsize=(12, 6))
ts_data[['매출', '비용']].plot()
plt.title('월별 매출 및 비용 추이')
plt.xlabel('날짜')
plt.ylabel('금액')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# 시프트 및 차분
# 시프트
ts_data['매출_전월'] = ts_data['매출'].shift(1)
print("매출과 전월 매출:")
print(ts_data[['매출', '매출_전월']])

# 차분
ts_data['매출_증감'] = ts_data['매출'].diff()
print("\n매출 증감:")
print(ts_data[['매출', '매출_증감']])

# 퍼센트 변화율
ts_data['매출_증감률'] = ts_data['매출'].pct_change() * 100
print("\n매출 증감률 (%):")
print(ts_data[['매출', '매출_증감률']])
```

```python
# 롤링 윈도우
# 3개월 이동 평균
ts_data['매출_3개월이동평균'] = ts_data['매출'].rolling(window=3).mean()
print("3개월 이동 평균:")
print(ts_data[['매출', '매출_3개월이동평균']])

# 이동 평균 시각화
plt.figure(figsize=(12, 6))
ts_data[['매출', '매출_3개월이동평균']].plot()
plt.title('매출 및 3개월 이동 평균')
plt.xlabel('날짜')
plt.ylabel('매출')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# 텍스트 데이터 처리
# 텍스트 데이터 생성
text_data = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '정지은', '최동욱'],
    '이메일': ['kim@example.com', 'lee@gmail.com', 'park@naver.com', 'jung@example.com', 'choi@gmail.com'],
    '주소': ['서울시 강남구', '부산시 해운대구', '서울시 마포구', '인천시 연수구', '경기도 성남시'],
    '취미': ['축구, 농구', '독서, 영화', '게임, 음악, 여행', '요리, 여행', '등산, 사진']
})

print("텍스트 데이터:")
print(text_data)
```

## 11. Scipy 연습

```python
# Scipy 임포트
from scipy import stats
import scipy.optimize as optimize

# 샘플 데이터 생성
np.random.seed(42)
group_a = np.random.normal(loc=75, scale=10, size=30)  # 평균 75, 표준편차 10인 정규분포
group_b = np.random.normal(loc=80, scale=10, size=30)  # 평균 80, 표준편차 10인 정규분포

df_stats = pd.DataFrame({
    'Group A': group_a,
    'Group B': group_b
})

print("통계 분석용 데이터:")
print(df_stats.head())
```

```python
# 기술 통계량
print("Group A 기술 통계량:")
print(stats.describe(group_a))
print("\nGroup B 기술 통계량:")
print(stats.describe(group_b))
```

```python
# t-검정
t_stat, p_value = stats.ttest_ind(group_a, group_b)
print("독립표본 t-검정 결과:")
print(f"t-통계량: {t_stat:.4f}")
print(f"p-값: {p_value:.4f}")
print(f"결론: 두 그룹은 {'통계적으로 유의한 차이가 있습니다' if p_value < 0.05 else '통계적으로 유의한 차이가 없습니다'}")
```

```python
# 상관 분석
# 상관계수 계산
correlation, p_value = stats.pearsonr(df_stats['Group A'], df_stats['Group B'])
print("Pearson 상관 분석 결과:")
print(f"상관계수: {correlation:.4f}")
print(f"p-값: {p_value:.4f}")
print(f"결론: 두 변수 간에는 {'통계적으로 유의한 상관관계가 있습니다' if p_value < 0.05 else '통계적으로 유의한 상관관계가 없습니다'}")

# 산점도 및 회귀선
plt.figure(figsize=(8, 6))
plt.scatter(df_stats['Group A'], df_stats['Group B'])
plt.title(f'Group A와 Group B의 상관관계 (r={correlation:.4f})')
plt.xlabel('Group A')
plt.ylabel('Group B')
plt.grid(True, alpha=0.3)

# 회귀선 추가
slope, intercept = np.polyfit(df_stats['Group A'], df_stats['Group B'], 1)
x = np.array([df_stats['Group A'].min(), df_stats['Group A'].max()])
plt.plot(x, slope * x + intercept, 'r--')
plt.show()
```

```python
# 곡선 피팅
# 데이터 생성
x = np.linspace(0, 10, 100)
y = 2 * np.sin(x) + 0.5 * x + np.random.normal(scale=0.5, size=len(x))

# 피팅 함수 정의
def func(x, a, b, c):
    return a * np.sin(x) + b * x + c

# 곡선 피팅
params, params_covariance = optimize.curve_fit(func, x, y, p0=[2, 0.5, 0])
print("곡선 피팅 결과:")
print(f"a = {params[0]:.4f}, b = {params[1]:.4f}, c = {params[2]:.4f}")

# 피팅 결과 시각화
plt.figure(figsize=(10, 6))
plt.scatter(x, y, label='데이터')
plt.plot(x, func(x, *params), 'r-', label='피팅: a*sin(x) + b*x + c')
plt.title('곡선 피팅 예제')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
