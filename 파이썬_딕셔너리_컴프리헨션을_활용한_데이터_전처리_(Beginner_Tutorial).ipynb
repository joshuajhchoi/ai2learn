* 특히 데이터 전처리 과정에서 사전을 빠르게 생성하거나 필터링하는 데 유용하게 활용할 수 있습니다.

###예제 1: 조건부 딕셔너리 컴프리헨션

**조건을 활용한 딕셔너리 컴프리헨션 예제입니다. 학생들의 점수가 담긴 사전에서 70점 이상인 항목만 골라내어 새로운 dictionary 만듭니다.**

```python
student_scores = {"철수": 85, "영희": 60, "민수": 90, "지영": 72}
# 70점 이상인 학생만 선택하여 새로운 딕셔너리 생성
passed_students = {name: score for name, score in student_scores.items() if score >= 70}
print(passed_students)
```

##### 설명: 첫 번째 예제에서는 딕셔너리 컴프리헨션에 조건식을 적용하는 방법을 보여드립니다. 학생점수(student_scores)라는 딕셔너리에 여러 학생들의 점수가 들어 있는데요.
이 중에서 점수가 70점 이상인 학생들만 골라 새로운 사전에 담고 있습니다. 딕셔너리 컴프리헨션의 끝에 if score >= 70 과 같은 조건을 추가하면 해당 조건을 만족하는 항목들만 포함됩니다.

이렇게 하면 한 줄의 코드로 필요한 데이터만 추려서 합격자 (passed_students)딕셔너리에 넣을 수 있습니다.

결과적으로 출력된 합격자 (passed_students)딕셔너리에는 성적이 70점 이상인 학생들만 이름과 점수가 들어 있죠.

이처럼 조건부 컴프리헨션을 사용하면 데이터를 필터링하면서도 매우 간결하게 딕셔너리를 만들 수 있습니다.

###예제 2: 중첩 루프를 사용하여 재분류하는 딕셔너리 컴프리헨션

####중첩된 for 루프를 활용한 딕셔너리 컴프리헨션 예제입니다. 과일과 채소 분류별로 항목 목록이 담긴 딕셔너리를 항목별 분류 딕셔너리로 변환합니다.

```python
items_by_category = { "Fruits": ["Apple", "Banana"], "Vegetables": ["Carrot", "Broccoli"] }
# 각 항목을 키로 하여 해당 항목의 분류를 값으로 갖는 딕셔너리 생성
item_to_category = { item: category for category, item_list in items_by_category.items() for item in item_list }
print(item_to_category)
```

#####두 번째 예제는 딕셔너리 컴프리헨션에서 이중 for 루프를 사용하는 방법입니다.

items_by_category라는 딕셔너리에 과일과 채소 같은 카테고리별로 여러 항목들이 리스트로 들어 있습니다.

이 데이터를 항목을 키로 하고 해당 항목의 분류를 밸류로 가지는 새로운 딕셔너리로 만들어 보겠습니다.

이를 위해 딕셔너리 컴프리헨션 안에 두 개의 for 문을 연달아 작성했습니다.

먼저 바깥쪽 for 문에서 각 분류 이름과 항목 목록을 가져오고, 이어서 안쪽 for 문에서 그 항목 목록 내부를 돌면서 항목 하나하나를 키로 추가합니다.

이렇게 하면 각 항목이 어떤 분류에 속하는지를 바로 찾아볼 수 있는 item_to_category 딕셔너리이 만들어집니다.

중첩된 포문으로된 컴프리헨션을 활용하여 계층적인 데이터를 재구성할 때 유용한 패턴입니다.


###예제 3: 리스트로부터 값 계산 딕셔너리 컴프리헨션

#####리스트의 각 요소로부터 값을 계산하여 딕셔너리를 만드는 예제입니다. 과일 이름 리스트가 주어지면, 각 과일 이름의 길이를 구해서 사전에 저장해봅니다.

```python
# 각 과일 이름의 길이를 계산하여 딕셔너리에 저장
fruits = ["apple", "banana", "pineapple", "watermelon"]
fruit_lengths = { fruit: len(fruit) for fruit in fruits }
print(fruit_lengths)
```

####세 번째 예제에서는 리스트로부터 새로운 값을 계산하여 딕셔너리를 채우는 방법을 보여줍니다.

fruits라는 리스트에 과일 이름들이 들어 있고, 우리는 각 과일 이름의 글자 수를 계산해서 그 결과를 딕셔너리에 담아볼 것입니다.

딕셔너리 컴프리헨션을 사용하면 리스트를 순회하면서 len(fruit)처럼 각 항목에 대한 계산을 바로 수행하고, 그 값을 해당 항목의 딕셔너리 값으로 저장할 수 있습니다.

이렇게 생성된 fruit_lengths 딕셔너리를 출력해보면 각 과일 이름과 그 길이를 한눈에 볼 수 있습니다.

이처럼 컴프리헨션을 활용하면 별도의 긴 for 문 없이도 리스트의 각 요소에 대한 특정 정보를 손쉽게 추출하여 딕셔너리에 담을 수 있습니다.

###예제 4: 딕셔너리 컴프리헨션으로 필터링 및 중복 제거

#### 리스트에서 중복된 항목을 제거하는 딕셔너리 컴프리헨션 예제입니다.

동일한 이름이 반복된 리스트를 처리하여 고유한 이름만 딕셔너에 키로 남깁니다.

```python
names_with_duplicates = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
# 딕셔너리 컴프리헨션으로 리스트의 중복을 제거
# 각 이름의 값으로 True를 저장 (존재 표시용 임시 값(Placeholder))
unique_names = { name: True for name in names_with_duplicates }
print(unique_names)
```

#####네 번째 예제에서는 딕셔너리 컴프리헨션을 활용하여 리스트의 중복된 항목을 제거해보겠습니다.

#####names_with_duplicates라는 리스트에 동일한 이름들이 여러 번 들어 있는데요. 이 리스트를 딕셔너리 컴프리헨션으로 처리하면, 딕셔너리의 키는 고유해야 하기 때문에 나중에 등장한 동일한 키 값이 이전 값을 덮어쓰게 됩니다.

#####그 결과 최종 생성된 unique_names 딕셔너리에는 각 이름이 한 번씩만 키로 남게 되죠.

####코드에서 각 이름의 값으로 True를 넣은 것은 그 이름이 존재한다는 표시(Place holder)일 뿐이고, 실제 값에는 특별한 의미가 없습니다.

#### 이렇게 딕셔너리를 사용하여 중복을 제거하면 리스트에서 고유한 값들만 빠르게 추려낼 수 있습니다.

#####딕셔너리 키의 고유성 특성을 이용해 데이터를 정리하는 유용한 트릭이며, 필요에 따라 이렇게 만들어진 키 목록을 활용하거나 list(unique_names.keys())처럼 키만 뽑아낼 수도 있습니다.

이번 실습을 통해 파이썬의 딕셔너리 컴프리헨션이 어떻게 데이터 전처리에 활용될 수 있는지 네 가지 사례로 알아보았습니다.

컴프리헨션을 사용하면 복잡한 반복문이나 조건문 로직을 한 줄로 묶어 표현할 수 있기 때문에, 코드가 훨씬 간결해지고 읽기 쉬워집니다.

특히 필터링, 형 변환, 구조 변환, 중복 제거와 같이 데이터를 가공하는 작업에서 유용하게 쓰일 수 있습니다.
