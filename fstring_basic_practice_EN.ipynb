# Python Bssic f-string Practice

This notebook contains basic examples for practicing Python's f-strings (formatted string literals). f-strings were introduced in Python 3.6 and allow you to embed variables and expressions directly within strings.

```python
# Inserting basic variables
name = "John"
age = 30

# Using f-strings
message = f"Hi, my name is {name} and I am {age} years old."
print(message)
```

```python
# Inserting multiple variables
first_name = "John"
last_name = "Doe"
job = "developer"
company = "Tech Corp"

introduction = f"My name is {first_name} {last_name}. I work as a {job} at {company}."
print(introduction)
```

```python
# Performing simple calculations inside an f-string
a = 10
b = 20

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} raised to the power of {b} = {a ** b}")
```

```python
# Using more complex expressions
x = 5
y = 10
z = 15

average = f"The average of the three numbers is {(x + y + z) / 3}"
print(average)

# Calling functions inside f-strings
result = f"Max value: {max(x, y, z)}, Min value: {min(x, y, z)}"
print(result)
```

```python
# Using string methods with f-strings
message = "hello, world!"

print(f"Original: {message}")
print(f"Uppercase: {message.upper()}")
print(f"Capitalized: {message.capitalize()}")
print(f"Title case: {message.title()}")
print(f"Position of 'world': {message.find('world')}")
```

```python
# Displaying percentages and scientific notation
value1 = 0.25
value2 = 0.75
value3 = 12345.6789

print(f"Percentage: {value1:.0%}")
print(f"Percentage (2 decimal places): {value2:.2%}")

print(f"Scientific notation: {value3:e}")
print(f"Scientific notation (2 decimal places): {value3:.2e}")
print(f"Scientific notation (uppercase): {value3:.2E}")
```

```python
# Formatting date and time objects
from datetime import date, time

birthday = date(1990, 5, 15)
print(f"Birthday: {birthday:%B %d, %Y}")
print(f"Birthday (weekday): {birthday:%A, %B %d, %Y}")

meeting_time = time(14, 30, 0)
print(f"Meeting time: {meeting_time:%I:%M %p}")
print(f"Meeting time (24-hour): {meeting_time:%H:%M}")
```

```python
# Conditional formatting with f-strings
users = [
    {"name": "Alice", "age": 25, "active": True},
    {"name": "Bob", "age": 35, "active": False},
    {"name": "Charlie", "age": 42, "active": True}
]

for user in users:
    status = "Active ✓" if user["active"] else "Inactive ✗"
    age_group = (
        "Young Adult" if user["age"] < 30
        else "Adult" if user["age"] < 40
        else "Middle-aged"
    )
    print(f"Name: {user['name']}, Age: {user['age']} ({age_group}), Status: {status}")
```

```python
# Combining list comprehension with f-strings
numbers = list(range(1, 11))

squares_cubes = [f"{num}: {num}² = {num**2}, {num}³ = {num**3}" for num in numbers]

for item in squares_cubes:
    print(item)
```

```python
# Conditional list comprehension with f-strings
result = [f"{num} is {'even' if num % 2 == 0 else 'odd'}." for num in range(1, 11)]
for r in result:
    print(r)

even_squares = [f"Square of {num}: {num**2}" for num in range(1, 21) if num % 2 == 0]
print("\nSquares of even numbers:")
for square in even_squares:
    print(square)
```

```python
# Using f-strings with a custom class
class Person:
    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job

    def __str__(self):
        return f"{self.name}, {self.age} years old, {self.job}"

    def introduce(self):
        return f"Hi! I'm {self.name}, {self.age} years old, and I work as a {self.job}."

    def is_adult(self):
        return self.age >= 18

# Creating objects
people = [
    Person("John Smith", 35, "Software Developer"),
    Person("Emma Johnson", 28, "Marketing Manager"),
    Person("Michael Brown", 17, "Student")
]

# Printing info with f-strings
for person in people:
    print(f"Info: {person}")
    print(f"Introduction: {person.introduce()}")
    print(f"Adult: {'Yes' if person.is_adult() else 'No'}")
    print()
```
