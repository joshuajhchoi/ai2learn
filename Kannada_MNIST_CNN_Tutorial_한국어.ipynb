# ** 2. Convolutional Neural Networks** <a id="100"></a>
<mark>[Return Contents](#0)

* 딥러닝에서 CNN (Convolutional Neural Network) 또는 ConvNet (Convolutional Neural Network)은 심층 신경 네트워크의 클래스로, 시각적 이미지 분석에 가장 일반적으로 적용됩니다.
* 이미지 및 비디오 인식, 추천 시스템, 이미지 분류, 의료 이미지 분석 및 자연어 처리 응용 프로그램이 있습니다.
*“Convolutional Neural Network”라는 이름은 네트워크가 Convolution이라는 수학적 연산을 사용함을 나타냅니다. 컨볼루션은 특별한 종류의 선형 연산입니다. 컨볼 루션 네트워크는 레이어 중 하나 이상에서 일반 행렬 곱셈 대신 컨볼 루션을 사용하는 단순한 신경망입니다.
* 컨볼루션 신경망은 하나 또는 여러 개의 컨볼 루션 층으로 구성되며 클래식 다층 피드 포워드 신경망에서와 같이 완전히 연결된 뉴런 레이어가 뒤 따릅니다.
* CNN의 숨겨진 계층은 일반적으로 회선 계층, 풀링 계층, 완전히 연결된 계층 및 정규화 계층으로 구성됩니다.

## **Convolution Operation** <a id="110"></a>



* 좌측은 "입력"신호 (또는 이미지)로, 중간(커널)은 입력 이미지에서 "필터"로 생각할 수 있고 이들이 세번째인 출력 이미지를 생성합니다 (따라서 컨볼루션은 두 개의 매트릭스를 입력으로 사용하여 생성합니다).


#### **Steps of convulation operation** <a id="111"></a>
1. 왼쪽 위의 하위 행렬의 요소곱을 택하고 필터
2. 모든 행과 열을 합하여 단일 값을 얻습니다.
3. 결과는 새 출력 행렬의 상단 모서리 셀 값입니다. (output(0,0))
4. 그런 다음 기본 이미지의 "S"열 위로 필터를 밀어 넣고 동일한 작업을 수행하십시오.  output(0,1) value가 생깁니다.

#### **Stride** <a id="112"></a>
* 슬라이딩 양은 stride (S) constant로 선언됩니다.  
* stride content는 변경 될 수 있으며 이는 출력 차원에 직접 영향을 미칩니다.
* $ l $ '번째 레이어에 대한 출력 매트릭스 차원 계산 :
\begin{equation}
n^{l} = \frac{n^{l-1} - f}{s}+1 \\
\text{output} =  n^l x n^l
\end{equation}
설명;
 * $l$ - layer number.
 * $n^{l}$ - horizontal/vertical dimension of output matrix.
 * $n^{l-1}$ -  horizontal/vertical dimension of input matrix.
 * $f$ - horizontal/vertical dimension of filter (kernel) matrix.
 * $s$ - stride constant. 다음 셀에서 출력 행렬을 계산하기 위해 커널이 오른쪽으로 얼마나 많은 열을 이동할지 결정하는 데 사용합니다.

### **An Example of Convolution Operation** <a id="113"></a>

    


* 위에서 볼 수 있듯이 5x5 이미지 매트릭스와 3x3 필터 매트릭스가 있습니다.
* stride (S) 파라미터를 1로 선택하면 출력 매트릭스는 3x3 매트릭스가됩니다.
* stride (S) 파라미터를 2로 선택하면 출력 매트릭스는 2x2 매트릭스가됩니다.

\begin{equation}
n^{l} = \frac{5 - 3}{1}+1 = 3\\
\text{output} =  3 x 3\\
\text{  }\\
n^{l} = \frac{5 - 3}{2}+1 = 2\\
\text{output} =  2 x 2\\
\end{equation}

* 처음 사진이 아래와 같이 컬러 (RGB) 인 경우 2 차원 및 3 차원도 사용할 수 있습니다. (6x6x3)
* 이 경우 필터 1을 3 개의 채널에서 세 개의 다른 값을 갖도록 변환하고 결과를 합산하여 단일 숫자를 얻습니다.
* 여기서 중요한 요소는 filter1이 모든 3 채널에 대해 동일한 값을 가지고 있다는 것입니다 (다른 색상으로 표시).
* 위의 예에서는 하나의 필터 만 사용하지만 실제 응용 프로그램에서는 아래와 같이 더 많은 필터가 사용됩니다.
* 출력 매트릭스 너비 (w)와 높이 (h) 크기는 일반적으로 응용 프로그램에서 동일하지만 다를 수도 있습니다.
* 따라서 출력 이미지는 $ w $ x $ h $ x $ c $ 차원의 행렬입니다.
* $ w $ 및 $ h $는 위의 방정식으로 계산되며 행렬의 채널 치수는 컨볼루션에 사용되는 필터 번호와 같습니다. 컨벌루션 연산 후 각 필터가 아래와 같이 매트릭스에서 다른 레이어를 생성한다고 상상할 수 있습니다.

### **Why do we use Filters?** <a id="114"></a>

* 이 질문에 먼저 답하기 위해서는 또 다른 일반적인 질문을해야합니다.
* 우리의 주요 목적은 무엇입니까?
* 이 커널에서 우리의 주요 목적은 필기 패턴을 인식 하는 것입니다.
* 당신은 고양이 얼굴 사진을 볼 때 어떻게 그것이 고양이인지 이해합니까? 얼굴에 확실한 뭐가 있어야합니까?
   * 눈 모양, 코 곡선, 콧수염 등과 같은 확실한 패턴
   * 우리의 뇌는 얼굴의 일부를 개별적으로 인식하고 결합하여 최종 결정을 내립니다. CNN 필터에 대해서도 동일한 설명이 허용됩니다.
* 각 필터에는 작업이 있습니다. 그들 중 일부는 가장자리를 감지하고 (예와 같이 Sobel 필터) 외곽선 일부만 감지합니다.
* 진화하는 네트워크의 딥 레이어에서 필터 작업이 더욱 복잡해집니다. 예를 들어, 네트워크의 첫 번째 레이어는 __horizontal lines__ 만 탐지 할 수있는 반면 네트워크의 더 깊은 레이어는 코나 눈을 탐지 할 수 있으며 __ 계속 진행중인 더 깊은 레이어는 사람의 얼굴까지도 탐지 할 수 있습니다 __
* 요약 필터는 이미지의 특징을 추출하는 데 사용됩니다. 피쳐 추출 후 추출 된 피쳐에 따라 분류를 위해 완전히 연결된 레이어를 사용합니다.


## **Padding** <a id="130"></a>

    

##### 패딩은 입력 행렬에 값을 대칭 적으로 추가하는 과정을 말합니다.



### **__Why do we use Padding?__** <a id="131"></a>

1. 가장자리의 정보는 한 번만 뒤 얽히고 가운데의 정보는 3x3 필터로 3번 뒤얽힙니다. 이로 인해 가장자리에 대한 정보가 손실됩니다. 패딩으로 인해 에지 정보가 여러 번 복잡해질 수 있습니다.
2. 컨볼루션 연산을 사용할 때마다 출력 매트릭스 크기가 작아집니다. 패딩은 입력 볼륨의 크기를 출력 볼륨에 유지해야 할 때 CNN 레이어를 설계 할 때 주로 사용됩니다 (** 동일 패딩 **).
    
### 패딩 프로세스를 사용하여 출력 차원 변경을 계산하는 방정식 <a id="132"></a>

\begin{equation}
n^{l} = \frac{n^{l-1} + 2p - f}{s}+1 \\
\text{output} =  n^l x n^l
\end{equation}


### **Types of Padding in Keras** <a id="133"></a>

* Valid : No padding
* Same  : 위의 방정식에서 $p$ 값을 선택하여 입력 행렬과 같은 차원의 출력 행렬을 얻습니다.

## **Pooling** <a id="150"></a>


* 풀링이란 매트릭스를 풀로 나누고 해당 풀을 나타내는 것으로서 하나의 값만 모든 풀에서 선택하는 프로세스를 말합니다.
* 컨볼 루션 작업에도 불구하고 풀링에는 필터를 사용하지 않으므로 가중치가 없습니다.
* 소그룹에서 하나의 값만 선택하는 것입니다.

### **Hyper Parameters of Pooling Operation** <a id="151"></a>
1. Pooling Size
2. Stride

* 위 그림에서의 컨볼루션 작업과 동일합니다. $2$x$2$ pooling size가 있고 stride = 2입니다. (sliding on matrix by 2 cell)

### **Why do we use Pooling?** <a id="152"></a>

1. 특징 추출에 대한 견고성을 얻기 위해;
     * 풀링은 선택한 값과 관련하여 불필요한 데이터를 삭제하여 모델이 과도하게 훈련되지 않도록합니다.
2. 계산 속도를 높이기 위해;
     * 계산 속도를 높이기 위해 표현의 크기를 줄입니다.

### **Types of Pooling in Keras** <a id="153"></a>

    
* Max Pooling: 위 그림과 같이 각 미니 그룹 (풀)에서 최대 값을 선택하는 작업
* Average Pooling: 각 미니 그룹에서 평균 값을 선택하는 작업

1. 특정 패턴을 포함하는 필터 매트릭스
2. Conv 레이어에서는 stride 수만큼 필터가 이미지 위로 이동하고 컨볼루션이 수행됩니다.
3. 3번째 과정의 결과로 얻은 행렬을 feature map이라고합니다. 여기서 이전 레이어의 매트릭스 영역에서 필터 패턴에 사용할 수있는 데이터의 양을 확인할 수 있습니다.
4. 수치 적으로 높은 값은 패턴대로 되고 있음을 나타냅니다.
5. feature map은 다음 CNN 계층에 대한 새로운 그림으로 간주 될 수 있습니다. 이제 새로운 필터로이 그림의 패턴을 감지 할 수 있습니다. 이것은 CNN의 두 번째 계층입니다. 이 로직으로 레이어 수를 늘릴 수 있습니다.

## **Batch Normalization** <a id="160"></a>


#### **What is Batch Norm?** <a id="161"></a>

* Batch Norm은 활성화를 조정하고 조정하여 숨겨진 레이어의 입력을 정규화하는 데 사용되는 방법입니다.
* 이 방법은 각 훈련 미니 배치에서 특정 층의 각 뉴런에 대해 정규화를 수행하는 것으로 구성됩니다.
* $1$ 레이어의 정규화 된 뉴런은 훈련 된 $\gamma$ 및 $\beta$ 매개 변수에 따라 문제에 대한 특정 평균 및 분산을 갖습니다.
* Batch Norm의 알고리즘은 다음과 같습니다.



#### **Why do we use Batch Norm?** <a id="161"></a>

* 인공 신경망에서 이전 레이어의 가중치 매개 변수가 변경되므로 훈련 중 각 레이어에 대한 입력 분포가 변경됩니다. 이로 인해 낮은 학습 속도와 신중한 매개 변수 초기화가 필요하므로 훈련 속도가 느려져 비선형 포화 상태의 모델을 훈련하기가 어렵습니다.

* Batch Norm 사용 이유
     1. 모델의 학습 속도를 높이기 위해.
     2. 훨씬 높은 학습 속도를 사용하고 initialization에 덜 주의 할 수 있게
     3. 네트워크를 정규화하려고 (side positive effect)
    

#### **Some notes about Batch Norm** <a id="162"></a>

* $\gamma$ 및 $\beta$ 매개 변수는 학습 가능한 매개 변수입니다. 따라서 그냥 초기화하고 backpropagation 또는 다른 최적화 알고리즘을 사용하면 가중치 매개 변수와 같이 최적화됩니다.
* 논문에서 covariate shifting을 호출하는 이전 레이어의 매개 변수가 향상 되었기 때문에 교육 중 각 레이어 입력의 분포를 변경합니다.
* Sergey Ioffe와 Christian Szegedy가 작성한 원본 논문은 여기 (https://arxiv.org/pdf/1502.03167.pdf)에서 읽을 수 있습니다.

## **Drop Out** <a id="180"></a>


* "Drop Out"은 효과적인 정규화 방법입니다.
* 훈련 중에 일부 수의 레이어 출력이 무작위로 무시되거나 "삭제됩니다". 이는 레이어 수를 다르게하고 이전 레이어에 대한 연결성을 가진 레이어처럼 보이게하고 처리하는 효과를줍니다. 실제로, 훈련 중 계층에 대한 각 업데이트는 구성된 계층의 다른 "보기"로 수행됩니다.
* __ 왜 "드롭 아웃"을 사용합니까? __ 대답은 과적합을 방지하는 것입니다 ...
* 밀도가 높은 완전 연결된 계층, 회선 계층 및 장기 단기 메모리 네트워크 계층과 같은 반복 계층과 같은 대부분의 계층 유형에 사용할 수 있습니다.

## **Building and Training a CNN Model** <a id="250"></a>

<hr>
* Keras Sequential Networks에는 모델 구축, 컴파일 및 피팅의 3 단계가 있습니다.

#### **Model - Build** <a id="251"></a>

* building 단계에서는 주로 모델의 아키텍처를 지정합니다.
* [Convolution] (# 110) 작업에 사용할 [Filter] (# 110) 크기 및 [Padding] (# 150) 유형을 결정하고 [Pooling] (# 150), Batch Normalization, Dropout, activation function layer에 빌드 섹션을 추가 할 수 있습니다

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(64, (3,3), padding='same', input_shape=(28, 28, 1)),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),
    tf.keras.layers.Conv2D(64,  (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),
    tf.keras.layers.Conv2D(64,  (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),

    tf.keras.layers.MaxPooling2D(2, 2),
    tf.keras.layers.Dropout(0.25),

    tf.keras.layers.Conv2D(128, (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),
    tf.keras.layers.Conv2D(128, (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),
    tf.keras.layers.Conv2D(128, (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),

    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Dropout(0.25),

    tf.keras.layers.Conv2D(256, (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),
    tf.keras.layers.LeakyReLU(alpha=0.1),
    tf.keras.layers.Conv2D(256, (3,3), padding='same'),
    tf.keras.layers.BatchNormalization(momentum=0.9, epsilon=1e-5, gamma_initializer="uniform"),##
    tf.keras.layers.LeakyReLU(alpha=0.1),

    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Dropout(0.25),


    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(256),
    tf.keras.layers.LeakyReLU(alpha=0.1),

    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dense(10, activation='softmax')
])
```

### **Model - Compile** <a id="252"></a>
    
* Compile Section에서는 모델 평가에 사용할 손실 함수, 최적화 알고리즘 및 메트릭을 지정합니다.

```python
optimizer = RMSprop(learning_rate=0.002,###########optimizer = RMSprop(learning_rate=0.0025,###########
    rho=0.9,
    momentum=0.1,
    epsilon=1e-07,
    centered=True,
    name='RMSprop')
model.compile(loss='categorical_crossentropy',
              optimizer=optimizer,
              metrics=['accuracy'])
```

#### **Compile ->Loss Function ->Categorical Crossentropy** <a id="253"></a>
    
* Categorical Crossentropy Loss Function은 다중 클래스 분류 문제에 대한 네트워크 예측과 목표 값 사이를 계산합니다.

* 다음 공식을 사용하여 loss를 계산합니다.
\begin{equation}
Loss = -\frac{1}{m}\sum_{i=1}^{m}\sum_{k=1}^{K} Y^i_klog(\hat{Y^i_k})(1-Y^i_k)log(1-\hat{Y^i_k}) \\
\end{equation}


여기서 $k$는 클래스를, $i$는 샘플 번호를, $\hat{Y_c}$는 예측 된 값, $Y_c$는 ground truth value, $m$은 배치의 샘플 번호이고 $K$는 총 클래스 수

* 로그를 사용합하는 이유는 교차 엔트로피 기능은 아래에 언급 된 것처럼 더 큰 차이를 더 크게, 더 적은 차이는 더 적게 페널티를 주기 때문입니다.



#### **Compile -> Optimizer** <a id="255"></a>

* 네트워크의 가중치를 최적화하려면 최적화 알고리즘을 선택해야합니다.
* gradient descent algorithm은 backpropagation에서 대표적으로 최적화 알고리즘에 사용되는 것이었습니다.
* 이제는 더 많은 알고리즘이 개발되어 사용되고 있습니다.
* 매표적으로 ADAGRAD, ADAM, ADAMAX, NADAM ,RMSPROP 등이지요.
* 이러한 알고리즘은 adaptive learning rate 나 global minimum을 얻기위한 momentum등을 사용합니다.
* 우측 링크를 통해 추가적인 학습을 하세요. [here](https://ruder.io/optimizing-gradient-descent/)by [Sebastian Ruder](https://ruder.io/optimizing-gradient-descent/).

* 아래에서 볼 수 있듯이  SGD (Stochastic Gradient Descent)는 새들 포인트를 벗어날 수 없으며, 고급 알고리즘은 다른 속도로 새들 포인트를 벗어날 수 있습니다

```python
batch_size = 1024
num_classes = 10
epochs = 40
```

#### **On-the-Fly Data Augmentation** <a id="257"></a>

* 이미지 데이터 세트의 분류 작업에서 데이터 확대는 모델의 일반화를 높이는 일반적인 방법입니다.

Keras의 ImageDataGenerator를 사용하면이 목표를 쉽게 처리 할 수 있습니다.
* [여기] (https://www.pyimagesearch.com/2019/07/08/keras-imagedatagenerator-and-data-augmentation/)는 [Adrian Rosebrock]이 작성한 데이터 확대 및 ImageDataGenerator에 대한 포괄적인 기사입니다 ( PyImageSearch의 저자 인 https://www.pyimagesearch.com/2019/07/08/keras-imagedatagenerator-and-data-augmentation/) 컴퓨터 비전에 대한 매우 유용한 웹 사이트입니다.

* 아래 코드에서 이미지의 일부 속성을 __ 변경하면 __ 데이터 세트에서 어떤 변화가 발생했는지 확인할 수 있습니다 __.
*이 관찰을 통해 선택해야 할 범위를 대략적으로 지정할 수 있습니다.

```python
# An observation code for our dataset
datagen_try = ImageDataGenerator(rotation_range=15,
                             width_shift_range = 0.15,
                             height_shift_range = 0.15,
                             shear_range = 0.15,
                             zoom_range = 0.4,)
# fit parameters from data
datagen_try.fit(x_train)
# configure batch size and retrieve one batch of images
for x_batch, y_batch in datagen_try.flow(x_train, y_train, batch_size=9):
	# create a grid of 3x3 images
	for i in range(0, 9):
		plt.subplot(330 + 1 + i)
		plt.imshow(x_batch[i].reshape(28, 28), cmap=plt.get_cmap('gray'))
	# show the plot
	plt.show()
	break
```

```python
datagen_train = ImageDataGenerator(rotation_range = 10,
                                   width_shift_range = 0.25,
                                   height_shift_range = 0.25,
                                   shear_range = 0.1,
                                   zoom_range = 0.4,
                                   horizontal_flip = False)

datagen_val = ImageDataGenerator()


step_train = x_train.shape[0] // batch_size
step_val = x_val.shape[0] // batch_size

learning_rate_reduction = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='loss',    # Quantity to be monitored.
    factor=0.25,       # Factor by which the learning rate will be reduced. new_lr = lr * factor
    patience=2,        # The number of epochs with no improvement after which learning rate will be reduced.
    verbose=1,         # 0: quiet - 1: update messages.
    mode="auto",       # {auto, min, max}. In min mode, lr will be reduced when the quantity monitored has stopped decreasing;
                       # in the max mode it will be reduced when the quantity monitored has stopped increasing;
                       # in auto mode, the direction is automatically inferred from the name of the monitored quantity.
    min_delta=0.0001,  # threshold for measuring the new optimum, to only focus on significant changes.
    cooldown=0,        # number of epochs to wait before resuming normal operation after learning rate (lr) has been reduced.
    min_lr=0.00001     # lower bound on the learning rate.
    )

es = EarlyStopping(monitor='val_loss', mode='min', verbose=1, patience=300, restore_best_weights=True)
```

### **Model - Fit** <a id="259"></a>


```python
history = model.fit_generator(datagen_train.flow(x_train, y_train, batch_size=batch_size),
                              steps_per_epoch=len(x_train)//batch_size,
                              epochs=epochs,
                              validation_data=(x_val, y_val),
                              validation_steps=50,
                              callbacks=[learning_rate_reduction, es],
                              verbose=2)
```

### **F1 Score Calculation** <a id="274"></a>


* 위에서 언급 한 것처럼 [정확도]는 모델의 성능에 대한 일반적인 아이디어를 제공하지만 모델의 추세에 대한 정보는 제공하지 않습니다.
* 분류 알고리즘에서 모델의 잘못된 예측을 분석하는 것이 중요합니다.
* false prediction에는 두 종류가 있습니다;

  * __Predicted as "1" but Ground Truth is "0" (False Positive)__
  * __Predicted as "0" but Ground Truth is "1" (False Negative)__

<a href="https://imgbb.com/"><img src="https://i.ibb.co/q99h0R6/f1.png" alt="f1" border="0"></a>
* F1 점수는 이러한 잘못된 예측 성능과 실제 긍정적 예측을 모두 고려하여 성공 성능 메트릭을 만듭니다.
* 문제가 2 개 이상의 클래스를 가지고 있기 때문에 각 클래스에 대한 F1 점수를 하나씩 계산했습니다.
* [여기] (https://towardsdatascience.com/multi-class-metrics-made-simple-part-ii-the-f1-score-ebe8b2c2ca1)는 여러 클래스에서 F1 점수를 계산하는 방법에 대한 좋은 기사입니다. 예제 지원됩니다.

```python
scores = np.zeros((10,3))
def calc_F1(num):
  TP = cm[num,num]
  FN = np.sum(cm[num,:])-cm[num,num]
  FP = np.sum(cm[:,num])-cm[num,num]
  precision = TP/(TP+FP)
  recall = TP/(TP+FN)
  F1_score = 2*(recall * precision) / (recall + precision)
  return precision, recall, F1_score
for i in range(10):
   precision, recall, F1_score = calc_F1(i)
   scores[i,:] = precision, recall, F1_score
scores_frame = pd.DataFrame(scores,columns=["Precision", "Recall", "F1 Score"], index=[list(range(0, 10))])
```

```python
f, ax = plt.subplots(figsize = (4,6))
ax.set_title('Number Scores')
sns.heatmap(scores_frame, annot=True, fmt=".3f", linewidths=0.5, cmap="PuBu", cbar=True, ax=ax)
bottom, top = ax.get_ylim()
plt.ylabel("")
ax.set_ylim(bottom + 0.5, top - 0.5)
plt.show()
```
